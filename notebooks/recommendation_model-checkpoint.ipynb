```python
import pandas as pd

# Load datasets
movies = pd.read_csv(r"C:\Users\M B Chinmay\Desktop\Zynema\data\movies.csv")
ratings = pd.read_csv(r"C:\Users\M B Chinmay\Desktop\Zynema\data\ratings.csv")

# Display first rows
print(movies.head())
print(ratings.head())
```

```python
movie_data = pd.merge(ratings, movies, on='movieId')

movie_data.head()
```

```python
# Take only first 100000 ratings
small_ratings = ratings.head(100000)

# Merge datasets
movie_data = pd.merge(small_ratings, movies, on='movieId')

# Keep only needed columns
movie_data_small = movie_data[['userId', 'title', 'rating']]

# Create user-movie matrix
user_movie_matrix = movie_data_small.pivot_table(
    index='userId',
    columns='title',
    values='rating',
    aggfunc='mean'
)

# Fill missing values
user_movie_matrix = user_movie_matrix.fillna(0)

# Display shape
print(user_movie_matrix.shape)

user_movie_matrix.head()
```

```python
from sklearn.metrics.pairwise import cosine_similarity

# Calculate movie similarity
movie_similarity = cosine_similarity(user_movie_matrix.T)

print(movie_similarity.shape)
```

```python
similarity_df = pd.DataFrame(
    movie_similarity,
    index=user_movie_matrix.columns,
    columns=user_movie_matrix.columns
)

similarity_df.head()
```

```python
def recommend_movies(movie_name, num_recommendations=5):

    similar_scores = similarity_df[movie_name].sort_values(
        ascending=False
    )

    return similar_scores[1:num_recommendations+1]
```

```python
recommend_movies("Toy Story (1995)")
```
